```python
#multilable text classification with ccna using augmented dataset
import json
import pandas as pd
import re
import numpy as np
from sklearn.preprocessing import MultiLabelBinarizer
import tensorflow as tf
from sklearn.metrics import classification_report, f1_score
import matplotlib.pyplot as plt
from tensorflow.keras.layers import Layer
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# 1. Data Loading
def load_and_preprocess(json_path, augment=False):
    with open(json_path, 'r', encoding='utf-8') as f:
        data = json.load(f)
    df = pd.DataFrame(data)

    # Extract techniques and clean text
    df["techniques"] = df["labels"].apply(lambda x: [label["technique"] for label in x])
    df["text"] = df["text"].apply(lambda x: re.sub(r"@\w+|https?://\S+|#[^\s]+", "", x))

    if augment:
        # Enhanced augmentation for rare classes
        rare_tech = ['Appeal to authority', 'Obfuscation, Intentional vagueness, Confusion', 'Repetition',
                    'Glittering generalities (Virtue)', 'Slogans']
        aug_samples = []

        def augment_text(text, technique_list):
            augmented = text
            # Technique-specific augmentations
            if 'Appeal to authority' in technique_list:
                augmented = re.sub(r'(\.|\?|!)', r' كما ذكر الخبير\1', augmented)
            if 'Obfuscation, Intentional vagueness, Confusion' in technique_list:
                augmented = re.sub(r'(\.|\?|!)', r' ربما يكون هذا صحيحًا\1', augmented)
            if 'Repetition' in technique_list:
                words = augmented.split()
                if len(words) > 3:
                    repeat_pos = np.random.randint(1, len(words)-1)
                    augmented = ' '.join(words[:repeat_pos] + [words[repeat_pos]] + words[repeat_pos:])
            if 'Glittering generalities (Virtue)' in technique_list:
                augmented = re.sub(r'(\.|\?|!)', r' وهذا من فضل الله\1', augmented)
            if 'Slogans' in technique_list:
                augmented = augmented + " الله أكبر"
            return augmented

        for _, row in df.iterrows():
            if any(t in row["techniques"] for t in rare_tech):
                # Create multiple augmented versions for rare classes
                for _ in range(3 if any(t in row["techniques"] for t in rare_tech[:3]) else 2):
                    aug_samples.append({
                        "text": augment_text(row["text"], row["techniques"]),
                        "techniques": row["techniques"]
                    })

        # augment some non-rare samples for balance
        non_rare_samples = df[~df["techniques"].apply(lambda x: any(t in x for t in rare_tech))].sample(frac=0.2)
        for _, row in non_rare_samples.iterrows():
            aug_samples.append({
                "text": row["text"] + " " + row["text"].split()[0],  # Simple duplication
                "techniques": row["techniques"]
            })

        df = pd.concat([df, pd.DataFrame(aug_samples)])

    return df

# 2. Data Preparation
train_df = load_and_preprocess("/content/drive/MyDrive/Colab Notebooks/dataset/all_aug_for_test.json", augment=True)
test_df = load_and_preprocess("/content/drive/MyDrive/Colab Notebooks/dataset/task2_test_gold_label_final_mod.json")

all_techniques = sorted(set(tech for sublist in train_df["techniques"] for tech in sublist))
mlb = MultiLabelBinarizer(classes=all_techniques)
mlb.fit(train_df["techniques"])

def prepare_dataset(df):
    labels_matrix = mlb.transform(df["techniques"])
    return pd.DataFrame({
        "text": df["text"],
        **{tech: labels_matrix[:, i] for i, tech in enumerate(mlb.classes_)}
    })

# 3. Advanced Text Cleaning
def clean_arabic_text(text):
    # Normalize Arabic characters more comprehensively
    text = re.sub(r"[إأآا]", "ا", text)
    text = re.sub(r"[ةه]", "ه", text)
    text = re.sub(r"[ىي]", "ي", text)
    text = re.sub(r"[ؤئ]", "ء", text)


    text = re.sub(r'[\u064B-\u065F]', '', text)  # Remove Arabic diacritics
    text = re.sub(r'[^\u0600-\u06FF\s،؛؟!.،]', '', text)  # Keep Arabic punctuation

    # Handle repeated punctuation
    text = re.sub(r'([!؟.])\1+', r'\1', text)

    # Normalize whitespace
    text = re.sub(r'\s+', ' ', text).strip()

    return text

train_processed = prepare_dataset(train_df)
test_processed = prepare_dataset(test_df)

for df in [train_processed, test_processed]:
    df["cleaned_text"] = df["text"].apply(clean_arabic_text)

# 4. Train and Val Split
X_train_full, y_train_full = train_processed["cleaned_text"], train_processed[all_techniques]
X_test, y_test = test_processed["cleaned_text"], test_processed[all_techniques]

test_size = 0.2
split_idx = int(len(X_train_full) * (1 - test_size))
X_train, X_val = X_train_full[:split_idx], X_train_full[split_idx:]
y_train, y_val = y_train_full[:split_idx], y_train_full[split_idx:]

# 5. Text Tokenization and Vectorization
MAX_NB_WORDS = 50000
MAX_SEQUENCE_LENGTH = 128
EMBEDDING_DIM = 300

tokenizer = Tokenizer(num_words=MAX_NB_WORDS, oov_token='<OOV>')
tokenizer.fit_on_texts(X_train)

word_index = tokenizer.word_index
print(f'Found {len(word_index)} unique tokens')

X_train_seq = tokenizer.texts_to_sequences(X_train)
X_val_seq = tokenizer.texts_to_sequences(X_val)
X_test_seq = tokenizer.texts_to_sequences(X_test)

X_train_pad = pad_sequences(X_train_seq, maxlen=MAX_SEQUENCE_LENGTH, padding='post', truncating='post')
X_val_pad = pad_sequences(X_val_seq, maxlen=MAX_SEQUENCE_LENGTH, padding='post', truncating='post')
X_test_pad = pad_sequences(X_test_seq, maxlen=MAX_SEQUENCE_LENGTH, padding='post', truncating='post')

# 6. Enhanced Attention Layer

class MultiHeadSelfAttention(tf.keras.layers.Layer):
    def __init__(self, num_heads=4, key_dim=64, **kwargs):
        super().__init__(**kwargs)
        self.num_heads = num_heads
        self.key_dim = key_dim
        self.attention = tf.keras.layers.MultiHeadAttention(
            num_heads=num_heads,
            key_dim=key_dim
        )
        self.layer_norm = tf.keras.layers.LayerNormalization()

    def build(self, input_shape):

        super().build(input_shape)

    def call(self, inputs):
        attention_output = self.attention(inputs, inputs)
        return self.layer_norm(inputs + attention_output)

    def get_config(self):
        config = super().get_config()
        config.update({
            'num_heads': self.num_heads,
            'key_dim': self.key_dim
        })
        return config


def build_enhanced_model():
    input_layer = tf.keras.layers.Input(shape=(MAX_SEQUENCE_LENGTH,))

    # Enhanced embedding with masking
    embedding_layer = tf.keras.layers.Embedding(
        input_dim=MAX_NB_WORDS,
        output_dim=EMBEDDING_DIM,
        input_length=MAX_SEQUENCE_LENGTH,
        mask_zero=True
    )(input_layer)

    # CNN branches with consistent dimensions
    conv_blocks = []
    for filter_size in [3, 4, 5]:
        conv = tf.keras.layers.Conv1D(
            filters=128,
            kernel_size=filter_size,
            padding='same',
            activation='relu'
        )(embedding_layer)
        conv = tf.keras.layers.BatchNormalization()(conv)
        conv = tf.keras.layers.MaxPooling1D(2)(conv)
        conv_blocks.append(conv)

    cnn_output = tf.keras.layers.concatenate(conv_blocks, axis=-1)

    # First BiLSTM layer
    lstm1 = tf.keras.layers.Bidirectional(
        tf.keras.layers.LSTM(128, return_sequences=True)
    )(cnn_output)
    lstm1 = tf.keras.layers.LayerNormalization()(lstm1)

    # Multi-head self-attention
    attention_output = MultiHeadSelfAttention(num_heads=4, key_dim=64)(lstm1)

    # Global pooling
    pooled_output = tf.keras.layers.GlobalAveragePooling1D()(attention_output)

    # Feature pyramid
    dense1 = tf.keras.layers.Dense(256, activation='swish')(pooled_output)
    dense1 = tf.keras.layers.BatchNormalization()(dense1)
    dense1 = tf.keras.layers.Dropout(0.5)(dense1)

    dense2 = tf.keras.layers.Dense(128, activation='swish')(dense1)
    dense2 = tf.keras.layers.BatchNormalization()(dense2)
    dense2 = tf.keras.layers.Dropout(0.3)(dense2)

    # Output layer
    output = tf.keras.layers.Dense(len(all_techniques), activation="sigmoid")(dense2)

    return tf.keras.Model(inputs=input_layer, outputs=output)


model = build_enhanced_model()

# 8. Enhanced Focal Loss with Class Balancing
class AdaptiveFocalLoss(tf.keras.losses.Loss):
    def __init__(self, class_frequencies, gamma_base=2.0, alpha=0.25, name='adaptive_focal_loss'):
        super().__init__(name=name)
        self.class_frequencies = class_frequencies
        self.gamma_base = gamma_base
        self.alpha = alpha

        # Convert to tensor with proper dtype
        self.gamma = {
            i: tf.constant(gamma_base * (1 + (1 - freq)), dtype=tf.float32)
            for i, freq in enumerate(class_frequencies)
        }
        self.alpha_tensor = tf.constant(alpha, dtype=tf.float32)

    def call(self, y_true, y_pred):
        y_true = tf.cast(y_true, tf.float32)
        y_pred = tf.cast(y_pred, tf.float32)

        pt = tf.where(tf.equal(y_true, 1), y_pred, 1 - y_pred)
        gamma_tensor = tf.zeros_like(y_true, dtype=tf.float32)

        for i in range(y_true.shape[1]):
            gamma_tensor = tf.where(
                tf.equal(y_true[:, i:i+1], 1),
                self.gamma[i],
                gamma_tensor
            )

        loss = -self.alpha_tensor * tf.pow(1.0 - pt, gamma_tensor) * tf.math.log(pt + 1e-7)
        return tf.reduce_mean(loss)

    def get_config(self):
        config = super().get_config()
        config.update({
            'class_frequencies': self.class_frequencies.tolist(),
            'gamma_base': self.gamma_base,
            'alpha': self.alpha
        })
        return config

    @classmethod
    def from_config(cls, config):
        config['class_frequencies'] = np.array(config['class_frequencies'])
        return cls(**config)


class_freq = y_train.mean(axis=0).values.astype(np.float32)
focal_loss = AdaptiveFocalLoss(class_freq)

# 9. Enhanced Class Weighting
class_counts = y_train.sum(axis=0)
total_samples = len(y_train)

# Enhanced class weighting formula
class_weights = {}
for i, count in enumerate(class_counts):
    if count == 0:
        class_weights[i] = 1.0  # Default weight for unseen classes
    else:
        # More aggressive weighting for rare classes
        weight = (total_samples / (len(class_counts) * count)) ** 0.5
        class_weights[i] = min(float(weight), 10.0)

# Special weights for specific techniques
special_weights = {
    'Appeal to authority': 8.0,
    'Obfuscation, Intentional vagueness, Confusion': 8.0,
    'Repetition': 6.0,
    'Glittering generalities (Virtue)': 6.0,
    'Slogans': 5.0,
    'Causal Oversimplification': 4.0,
    'Black-and-white Fallacy/Dictatorship': 4.0
}

for tech, weight in special_weights.items():
    if tech in all_techniques:
        class_weights[all_techniques.index(tech)] = weight

print("\nClass Weights:")
for tech, weight in zip(all_techniques, [class_weights[i] for i in range(len(all_techniques))]):
    print(f"{tech:>50}: {weight:.2f}")

# 10. F1 Score Metric
class F1MicroScore(tf.keras.metrics.Metric):
    def __init__(self, name='f1_micro', **kwargs):
        super().__init__(name=name, **kwargs)
        self.precision = tf.keras.metrics.Precision(thresholds=0.5)
        self.recall = tf.keras.metrics.Recall(thresholds=0.5)

    def update_state(self, y_true, y_pred, sample_weight=None):
        self.precision.update_state(y_true, y_pred)
        self.recall.update_state(y_true, y_pred)

    def result(self):
        p = self.precision.result()
        r = self.recall.result()
        return 2 * ((p * r) / (p + r + 1e-6))

    def reset_state(self):
        self.precision.reset_state()
        self.recall.reset_state()

# 11. Model Compilation and Training
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss=focal_loss,
    metrics=[
        tf.keras.metrics.AUC(name='auc', multi_label=True),
        tf.keras.metrics.Precision(name='precision', thresholds=0.5),
        tf.keras.metrics.Recall(name='recall', thresholds=0.5),
        F1MicroScore()
    ]
)

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor='val_f1_micro',
    patience=10,
    mode='max',
    restore_best_weights=True
)

reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=3,
    min_lr=1e-6
)

history = model.fit(
    X_train_pad,
    y_train,
    validation_data=(X_val_pad, y_val),
    batch_size=32,
    epochs=50,
    callbacks=[early_stopping, reduce_lr],
    class_weight=class_weights
)

# 12. Threshold Optimization
def optimize_thresholds(y_true, y_pred_probs):
    thresholds = []
    for i in range(y_pred_probs.shape[1]):
        if y_true[:, i].sum() == 0:
            thresholds.append(0.3)
            continue

        # Coarse search
        best_t, best_f1 = 0.3, 0
        for t in np.linspace(0.1, 0.5, 20):
            f1 = f1_score(y_true[:, i], y_pred_probs[:, i] > t, zero_division=0)
            if f1 > best_f1:
                best_t, best_f1 = t, f1

        # Fine search
        for t in np.linspace(max(0.1, best_t-0.05), min(0.5, best_t+0.05), 50):
            f1 = f1_score(y_true[:, i], y_pred_probs[:, i] > t, zero_division=0)
            if f1 > best_f1:
                best_t, best_f1 = t, f1
        thresholds.append(best_t)

    # Manual adjustments
    adjustments = {
        'Appeal to authority': 0.50,
        'Smears': 0.45,
        'Exaggeration/Minimisation': 0.45,
        'Flag-waving': 0.40,
        'Obfuscation, Intentional vagueness, Confusion': 0.35
    }
    for tech, thresh in adjustments.items():
        if tech in all_techniques:
            thresholds[all_techniques.index(tech)] = thresh

    return thresholds

val_pred_probs = model.predict(X_val_pad)
optimal_thresholds = optimize_thresholds(y_val.values, val_pred_probs)

# 13. Hybrid Prediction with Rules
def apply_hybrid_rules(texts, pred_probs, thresholds):
    rules = {
        'Appeal to authority': (
            r'(?:خبير|مختص|سلطة|توصية)(?:\s+\S+){0,3}\s+(?:قال|أكد|ذكر|صرح)',
            0.65,
            0.6
        ),
        'Obfuscation, Intentional vagueness, Confusion': (
            r'(?:بعض|عديد|معظم|قطاع)\s+(?:من|منهم|لها)\b|\bربما\b|يحتمل|قد يكون',
            0.55,
            0.4
        ),
        'Repetition': (
            r'(\b\w+\b)(?:\W+\1){2,}',
            0.5,
            0.4
        ),
        'Smears': (
            r'\b(?:فاسد|لص|سارق|نهب|خائن)\b(?:\s+\S+){0,2}\s+(?:اتهام|اتهم|وصف)',
            0.65,
            0.5
        ),
        'Flag-waving': (
            r'(?:وطن|أمة|شرف|انتماء)\s+(?:فخر|تضحيات|دماء|واجب)',
            0.7,
            0.5
        )
    }

    final_preds = np.zeros_like(pred_probs)
    tech_index = {tech:i for i, tech in enumerate(all_techniques)}

    for tech, (pattern, boost, max_thresh) in rules.items():
        if tech in tech_index:  # Only apply if technique exists in our classes
            i = tech_index[tech]
            compiled = re.compile(pattern)
            tech_thresh = min(thresholds[i] * boost, max_thresh)
            for j, text in enumerate(texts):
                if compiled.search(text):
                    final_preds[j,i] = int(pred_probs[j,i] > tech_thresh)
                else:
                    final_preds[j,i] = int(pred_probs[j,i] > thresholds[i])

    # Default thresholds for other classes
    for i, tech in enumerate(all_techniques):
        if tech not in rules:
            final_preds[:,i] = (pred_probs[:,i] > thresholds[i]).astype(int)

    # Ensure minimum one prediction per text
    for j in range(len(texts)):
        if not final_preds[j].any():
            final_preds[j,np.argmax(pred_probs[j])] = 1

    return final_preds

# 14. Final Evaluation
test_pred_probs = model.predict(X_test_pad)
test_preds = apply_hybrid_rules(X_test.tolist(), test_pred_probs, optimal_thresholds)

test_classes_present = [tech for tech in all_techniques if y_test[tech].sum() > 0]
test_class_indices = [all_techniques.index(tech) for tech in test_classes_present]

print("\nOptimized Thresholds:")
for tech, threshold in zip(all_techniques, optimal_thresholds):
    print(f"{tech:>50}: {threshold:.3f}")

print("\nEnhanced Test Performance:")
print(classification_report(
    y_test[test_classes_present],
    test_preds[:, test_class_indices],
    target_names=test_classes_present,
    zero_division=0
))

f1_micro = f1_score(y_test, test_preds, average='micro', zero_division=0)
print("F1-micro:", f1_micro)

# Visualization
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.plot(history.history['loss'], label='Train')
plt.plot(history.history['val_loss'], label='Validation')
plt.title('Loss')
plt.legend()

plt.subplot(1, 3, 2)
plt.plot(history.history['precision'], label='Train Precision')
plt.plot(history.history['val_precision'], label='Val Precision')
plt.title('Precision')
plt.legend()

plt.subplot(1, 3, 3)
plt.plot(history.history['recall'], label='Train Recall')
plt.plot(history.history['val_recall'], label='Val Recall')
plt.title('Recall')
plt.legend()
plt.tight_layout()
plt.show()

# 15. Save the Model
import os
import pickle

model_dir = "/content/drive/MyDrive/Colab Notebooks/models/enhanced_cnn_lstm_attention"
os.makedirs(model_dir, exist_ok=True)

# Save model
model.save(os.path.join(model_dir, "model.keras"))

# Save tokenizer
with open(os.path.join(model_dir, "tokenizer.pkl"), 'wb') as handle:
    pickle.dump(tokenizer, handle)

# Save metadata
metadata = {
    "mlb": mlb,
    "optimal_thresholds": optimal_thresholds,
    "all_techniques": all_techniques,
    "class_frequencies": class_freq,
    "max_sequence_length": MAX_SEQUENCE_LENGTH
}
with open(os.path.join(model_dir, "metadata.pkl"), 'wb') as handle:
    pickle.dump(metadata, handle)

print(f"Model and assets saved to: {model_dir}")
```

```python
#multilable text classification with ccna using original dataset
import json
import pandas as pd
import re
import numpy as np
from sklearn.preprocessing import MultiLabelBinarizer
import tensorflow as tf
from sklearn.metrics import classification_report, f1_score
import matplotlib.pyplot as plt
from tensorflow.keras.layers import Layer
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# 1. Data Loading
def load_and_preprocess(json_path, augment=False):
    with open(json_path, 'r', encoding='utf-8') as f:
        data = json.load(f)
    df = pd.DataFrame(data)

    # Extract techniques and clean text
    df["techniques"] = df["labels"].apply(lambda x: [label["technique"] for label in x])
    df["text"] = df["text"].apply(lambda x: re.sub(r"@\w+|https?://\S+|#[^\s]+", "", x))

    if augment:
        # Enhanced augmentation for rare classes
        rare_tech = ['Appeal to authority', 'Obfuscation, Intentional vagueness, Confusion', 'Repetition',
                    'Glittering generalities (Virtue)', 'Slogans']
        aug_samples = []

        def augment_text(text, technique_list):
            augmented = text
            # Technique-specific augmentations
            if 'Appeal to authority' in technique_list:
                augmented = re.sub(r'(\.|\?|!)', r' كما ذكر الخبير\1', augmented)
            if 'Obfuscation, Intentional vagueness, Confusion' in technique_list:
                augmented = re.sub(r'(\.|\?|!)', r' ربما يكون هذا صحيحًا\1', augmented)
            if 'Repetition' in technique_list:
                words = augmented.split()
                if len(words) > 3:
                    repeat_pos = np.random.randint(1, len(words)-1)
                    augmented = ' '.join(words[:repeat_pos] + [words[repeat_pos]] + words[repeat_pos:])
            if 'Glittering generalities (Virtue)' in technique_list:
                augmented = re.sub(r'(\.|\?|!)', r' وهذا من فضل الله\1', augmented)
            if 'Slogans' in technique_list:
                augmented = augmented + " الله أكبر"
            return augmented

        for _, row in df.iterrows():
            if any(t in row["techniques"] for t in rare_tech):
                # Create multiple augmented versions for rare classes
                for _ in range(3 if any(t in row["techniques"] for t in rare_tech[:3]) else 2):
                    aug_samples.append({
                        "text": augment_text(row["text"], row["techniques"]),
                        "techniques": row["techniques"]
                    })

        # augment some non-rare samples for balance
        non_rare_samples = df[~df["techniques"].apply(lambda x: any(t in x for t in rare_tech))].sample(frac=0.2)
        for _, row in non_rare_samples.iterrows():
            aug_samples.append({
                "text": row["text"] + " " + row["text"].split()[0],  # Simple duplication
                "techniques": row["techniques"]
            })

        df = pd.concat([df, pd.DataFrame(aug_samples)])

    return df

# 2. Data Preparation
train_df = load_and_preprocess("/content/drive/MyDrive/Colab Notebooks/dataset/task2_train.json", augment=True)
test_df = load_and_preprocess("/content/drive/MyDrive/Colab Notebooks/dataset/task2_test_gold_label_final_mod.json")

all_techniques = sorted(set(tech for sublist in train_df["techniques"] for tech in sublist))
mlb = MultiLabelBinarizer(classes=all_techniques)
mlb.fit(train_df["techniques"])

def prepare_dataset(df):
    labels_matrix = mlb.transform(df["techniques"])
    return pd.DataFrame({
        "text": df["text"],
        **{tech: labels_matrix[:, i] for i, tech in enumerate(mlb.classes_)}
    })

# 3. Advanced Text Cleaning
def clean_arabic_text(text):
    # Normalize Arabic characters more comprehensively
    text = re.sub(r"[إأآا]", "ا", text)
    text = re.sub(r"[ةه]", "ه", text)
    text = re.sub(r"[ىي]", "ي", text)
    text = re.sub(r"[ؤئ]", "ء", text)

    # Remove diacritics
    text = re.sub(r'[\u064B-\u065F]', '', text)  # Remove Arabic diacritics
    text = re.sub(r'[^\u0600-\u06FF\s،؛؟!.،]', '', text)  # Keep Arabic punctuation

    # Handle repeated punctuation
    text = re.sub(r'([!؟.])\1+', r'\1', text)

    # Normalize whitespace
    text = re.sub(r'\s+', ' ', text).strip()

    return text

train_processed = prepare_dataset(train_df)
test_processed = prepare_dataset(test_df)

for df in [train_processed, test_processed]:
    df["cleaned_text"] = df["text"].apply(clean_arabic_text)

# 4. Train/Val Split
X_train_full, y_train_full = train_processed["cleaned_text"], train_processed[all_techniques]
X_test, y_test = test_processed["cleaned_text"], test_processed[all_techniques]

test_size = 0.2
split_idx = int(len(X_train_full) * (1 - test_size))
X_train, X_val = X_train_full[:split_idx], X_train_full[split_idx:]
y_train, y_val = y_train_full[:split_idx], y_train_full[split_idx:]

# 5. Text Tokenization and Vectorization
MAX_NB_WORDS = 50000
MAX_SEQUENCE_LENGTH = 128
EMBEDDING_DIM = 300

tokenizer = Tokenizer(num_words=MAX_NB_WORDS, oov_token='<OOV>')
tokenizer.fit_on_texts(X_train)

word_index = tokenizer.word_index
print(f'Found {len(word_index)} unique tokens')

X_train_seq = tokenizer.texts_to_sequences(X_train)
X_val_seq = tokenizer.texts_to_sequences(X_val)
X_test_seq = tokenizer.texts_to_sequences(X_test)

X_train_pad = pad_sequences(X_train_seq, maxlen=MAX_SEQUENCE_LENGTH, padding='post', truncating='post')
X_val_pad = pad_sequences(X_val_seq, maxlen=MAX_SEQUENCE_LENGTH, padding='post', truncating='post')
X_test_pad = pad_sequences(X_test_seq, maxlen=MAX_SEQUENCE_LENGTH, padding='post', truncating='post')

# 6. Enhanced Attention Layer

class MultiHeadSelfAttention(tf.keras.layers.Layer):
    def __init__(self, num_heads=4, key_dim=64, **kwargs):
        super().__init__(**kwargs)
        self.num_heads = num_heads
        self.key_dim = key_dim
        self.attention = tf.keras.layers.MultiHeadAttention(
            num_heads=num_heads,
            key_dim=key_dim
        )
        self.layer_norm = tf.keras.layers.LayerNormalization()

    def build(self, input_shape):

        super().build(input_shape)

    def call(self, inputs):
        attention_output = self.attention(inputs, inputs)
        return self.layer_norm(inputs + attention_output)

    def get_config(self):
        config = super().get_config()
        config.update({
            'num_heads': self.num_heads,
            'key_dim': self.key_dim
        })
        return config


def build_enhanced_model():
    input_layer = tf.keras.layers.Input(shape=(MAX_SEQUENCE_LENGTH,))

    # Enhanced embedding with masking
    embedding_layer = tf.keras.layers.Embedding(
        input_dim=MAX_NB_WORDS,
        output_dim=EMBEDDING_DIM,
        input_length=MAX_SEQUENCE_LENGTH,
        mask_zero=True
    )(input_layer)


    conv_blocks = []
    for filter_size in [3, 4, 5]:
        conv = tf.keras.layers.Conv1D(
            filters=128,
            kernel_size=filter_size,
            padding='same',
            activation='relu'
        )(embedding_layer)
        conv = tf.keras.layers.BatchNormalization()(conv)
        conv = tf.keras.layers.MaxPooling1D(2)(conv)
        conv_blocks.append(conv)

    cnn_output = tf.keras.layers.concatenate(conv_blocks, axis=-1)

    # First BiLSTM layer
    lstm1 = tf.keras.layers.Bidirectional(
        tf.keras.layers.LSTM(128, return_sequences=True)
    )(cnn_output)
    lstm1 = tf.keras.layers.LayerNormalization()(lstm1)

    # Multi-head self-attention
    attention_output = MultiHeadSelfAttention(num_heads=4, key_dim=64)(lstm1)

    # Global pooling
    pooled_output = tf.keras.layers.GlobalAveragePooling1D()(attention_output)

    # Feature pyramid
    dense1 = tf.keras.layers.Dense(256, activation='swish')(pooled_output)
    dense1 = tf.keras.layers.BatchNormalization()(dense1)
    dense1 = tf.keras.layers.Dropout(0.5)(dense1)

    dense2 = tf.keras.layers.Dense(128, activation='swish')(dense1)
    dense2 = tf.keras.layers.BatchNormalization()(dense2)
    dense2 = tf.keras.layers.Dropout(0.3)(dense2)

    # Output layer
    output = tf.keras.layers.Dense(len(all_techniques), activation="sigmoid")(dense2)

    return tf.keras.Model(inputs=input_layer, outputs=output)


model = build_enhanced_model()

# 8. Enhanced Focal Loss with Class Balancing
class AdaptiveFocalLoss(tf.keras.losses.Loss):
    def __init__(self, class_frequencies, gamma_base=2.0, alpha=0.25, name='adaptive_focal_loss'):
        super().__init__(name=name)
        self.class_frequencies = class_frequencies
        self.gamma_base = gamma_base
        self.alpha = alpha

        # Convert to tensor with proper dtype
        self.gamma = {
            i: tf.constant(gamma_base * (1 + (1 - freq)), dtype=tf.float32)
            for i, freq in enumerate(class_frequencies)
        }
        self.alpha_tensor = tf.constant(alpha, dtype=tf.float32)

    def call(self, y_true, y_pred):
        y_true = tf.cast(y_true, tf.float32)
        y_pred = tf.cast(y_pred, tf.float32)

        pt = tf.where(tf.equal(y_true, 1), y_pred, 1 - y_pred)
        gamma_tensor = tf.zeros_like(y_true, dtype=tf.float32)

        for i in range(y_true.shape[1]):
            gamma_tensor = tf.where(
                tf.equal(y_true[:, i:i+1], 1),
                self.gamma[i],
                gamma_tensor
            )

        loss = -self.alpha_tensor * tf.pow(1.0 - pt, gamma_tensor) * tf.math.log(pt + 1e-7)
        return tf.reduce_mean(loss)

    def get_config(self):
        config = super().get_config()
        config.update({
            'class_frequencies': self.class_frequencies.tolist(),
            'gamma_base': self.gamma_base,
            'alpha': self.alpha
        })
        return config

    @classmethod
    def from_config(cls, config):
        config['class_frequencies'] = np.array(config['class_frequencies'])
        return cls(**config)


class_freq = y_train.mean(axis=0).values.astype(np.float32)
focal_loss = AdaptiveFocalLoss(class_freq)

# 9. Enhanced Class Weighting
class_counts = y_train.sum(axis=0)
total_samples = len(y_train)

# Enhanced class weighting formula
class_weights = {}
for i, count in enumerate(class_counts):
    if count == 0:
        class_weights[i] = 1.0  # Default weight for unseen classes
    else:
        # More aggressive weighting for rare classes
        weight = (total_samples / (len(class_counts) * count)) ** 0.5
        class_weights[i] = min(float(weight), 10.0)  # Cap at 10.0

# Special weights for specific techniques
special_weights = {
    'Appeal to authority': 8.0,
    'Obfuscation, Intentional vagueness, Confusion': 8.0,
    'Repetition': 6.0,
    'Glittering generalities (Virtue)': 6.0,
    'Slogans': 5.0,
    'Causal Oversimplification': 4.0,
    'Black-and-white Fallacy/Dictatorship': 4.0
}

for tech, weight in special_weights.items():
    if tech in all_techniques:
        class_weights[all_techniques.index(tech)] = weight

print("\nClass Weights:")
for tech, weight in zip(all_techniques, [class_weights[i] for i in range(len(all_techniques))]):
    print(f"{tech:>50}: {weight:.2f}")

# 10. F1 Score Metric
class F1MicroScore(tf.keras.metrics.Metric):
    def __init__(self, name='f1_micro', **kwargs):
        super().__init__(name=name, **kwargs)
        self.precision = tf.keras.metrics.Precision(thresholds=0.5)
        self.recall = tf.keras.metrics.Recall(thresholds=0.5)

    def update_state(self, y_true, y_pred, sample_weight=None):
        self.precision.update_state(y_true, y_pred)
        self.recall.update_state(y_true, y_pred)

    def result(self):
        p = self.precision.result()
        r = self.recall.result()
        return 2 * ((p * r) / (p + r + 1e-6))

    def reset_state(self):
        self.precision.reset_state()
        self.recall.reset_state()

# 11. Model Compilation and Training
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss=focal_loss,
    metrics=[
        tf.keras.metrics.AUC(name='auc', multi_label=True),
        tf.keras.metrics.Precision(name='precision', thresholds=0.5),
        tf.keras.metrics.Recall(name='recall', thresholds=0.5),
        F1MicroScore()
    ]
)

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor='val_f1_micro',
    patience=10,
    mode='max',
    restore_best_weights=True
)

reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=3,
    min_lr=1e-6
)

history = model.fit(
    X_train_pad,
    y_train,
    validation_data=(X_val_pad, y_val),
    batch_size=32,
    epochs=50,
    callbacks=[early_stopping, reduce_lr],
    class_weight=class_weights
)

# 12. Threshold Optimization
def optimize_thresholds(y_true, y_pred_probs):
    thresholds = []
    for i in range(y_pred_probs.shape[1]):
        if y_true[:, i].sum() == 0:
            thresholds.append(0.3)
            continue

        # Coarse search
        best_t, best_f1 = 0.3, 0
        for t in np.linspace(0.1, 0.5, 20):
            f1 = f1_score(y_true[:, i], y_pred_probs[:, i] > t, zero_division=0)
            if f1 > best_f1:
                best_t, best_f1 = t, f1

        # Fine search
        for t in np.linspace(max(0.1, best_t-0.05), min(0.5, best_t+0.05), 50):
            f1 = f1_score(y_true[:, i], y_pred_probs[:, i] > t, zero_division=0)
            if f1 > best_f1:
                best_t, best_f1 = t, f1
        thresholds.append(best_t)

    # Manual adjustments
    adjustments = {
        'Appeal to authority': 0.50,
        'Smears': 0.45,
        'Exaggeration/Minimisation': 0.45,
        'Flag-waving': 0.40,
        'Obfuscation, Intentional vagueness, Confusion': 0.35
    }
    for tech, thresh in adjustments.items():
        if tech in all_techniques:
            thresholds[all_techniques.index(tech)] = thresh

    return thresholds

val_pred_probs = model.predict(X_val_pad)
optimal_thresholds = optimize_thresholds(y_val.values, val_pred_probs)

# 13. Hybrid Prediction with Rules
def apply_hybrid_rules(texts, pred_probs, thresholds):
    rules = {
        'Appeal to authority': (
            r'(?:خبير|مختص|سلطة|توصية)(?:\s+\S+){0,3}\s+(?:قال|أكد|ذكر|صرح)',
            0.65,
            0.6
        ),
        'Obfuscation, Intentional vagueness, Confusion': (
            r'(?:بعض|عديد|معظم|قطاع)\s+(?:من|منهم|لها)\b|\bربما\b|يحتمل|قد يكون',
            0.55,
            0.4
        ),
        'Repetition': (
            r'(\b\w+\b)(?:\W+\1){2,}',
            0.5,
            0.4
        ),
        'Smears': (
            r'\b(?:فاسد|لص|سارق|نهب|خائن)\b(?:\s+\S+){0,2}\s+(?:اتهام|اتهم|وصف)',
            0.65,
            0.5
        ),
        'Flag-waving': (
            r'(?:وطن|أمة|شرف|انتماء)\s+(?:فخر|تضحيات|دماء|واجب)',
            0.7,
            0.5
        )
    }

    final_preds = np.zeros_like(pred_probs)
    tech_index = {tech:i for i, tech in enumerate(all_techniques)}

    for tech, (pattern, boost, max_thresh) in rules.items():
        if tech in tech_index:  # Only apply if technique exists in our classes
            i = tech_index[tech]
            compiled = re.compile(pattern)
            tech_thresh = min(thresholds[i] * boost, max_thresh)
            for j, text in enumerate(texts):
                if compiled.search(text):
                    final_preds[j,i] = int(pred_probs[j,i] > tech_thresh)
                else:
                    final_preds[j,i] = int(pred_probs[j,i] > thresholds[i])

    # Default thresholds for other classes
    for i, tech in enumerate(all_techniques):
        if tech not in rules:
            final_preds[:,i] = (pred_probs[:,i] > thresholds[i]).astype(int)

    # Ensure minimum one prediction per text
    for j in range(len(texts)):
        if not final_preds[j].any():
            final_preds[j,np.argmax(pred_probs[j])] = 1

    return final_preds

# 14. Final Evaluation
test_pred_probs = model.predict(X_test_pad)
test_preds = apply_hybrid_rules(X_test.tolist(), test_pred_probs, optimal_thresholds)

test_classes_present = [tech for tech in all_techniques if y_test[tech].sum() > 0]
test_class_indices = [all_techniques.index(tech) for tech in test_classes_present]

print("\nOptimized Thresholds:")
for tech, threshold in zip(all_techniques, optimal_thresholds):
    print(f"{tech:>50}: {threshold:.3f}")

print("\nEnhanced Test Performance:")
print(classification_report(
    y_test[test_classes_present],
    test_preds[:, test_class_indices],
    target_names=test_classes_present,
    zero_division=0
))

f1_micro = f1_score(y_test, test_preds, average='micro', zero_division=0)
print("F1-micro:", f1_micro)

# Visualization
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.plot(history.history['loss'], label='Train')
plt.plot(history.history['val_loss'], label='Validation')
plt.title('Loss')
plt.legend()

plt.subplot(1, 3, 2)
plt.plot(history.history['precision'], label='Train Precision')
plt.plot(history.history['val_precision'], label='Val Precision')
plt.title('Precision')
plt.legend()

plt.subplot(1, 3, 3)
plt.plot(history.history['recall'], label='Train Recall')
plt.plot(history.history['val_recall'], label='Val Recall')
plt.title('Recall')
plt.legend()
plt.tight_layout()
plt.show()

# 15. Save the Model
import os
import pickle

model_dir = "/content/drive/MyDrive/Colab Notebooks/models/enhanced_cnn_lstm_attention_original"
os.makedirs(model_dir, exist_ok=True)

# Save model
model.save(os.path.join(model_dir, "model.keras"))

# Save tokenizer
with open(os.path.join(model_dir, "tokenizer.pkl"), 'wb') as handle:
    pickle.dump(tokenizer, handle)

# Save metadata
metadata = {
    "mlb": mlb,
    "optimal_thresholds": optimal_thresholds,
    "all_techniques": all_techniques,
    "class_frequencies": class_freq,
    "max_sequence_length": MAX_SEQUENCE_LENGTH
}
with open(os.path.join(model_dir, "metadata.pkl"), 'wb') as handle:
    pickle.dump(metadata, handle)

print(f"Model and assets saved to: {model_dir}")
```
