```python
#multilable text classification bert_pos_ner using augmented dataset
# Import all required libraries
import json
import pandas as pd
import re
import numpy as np
from sklearn.preprocessing import MultiLabelBinarizer
from transformers import AutoTokenizer, TFAutoModel
import tensorflow as tf
from sklearn.metrics import classification_report, f1_score
import matplotlib.pyplot as plt
import stanza
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from scipy.sparse import hstack
import os
import joblib

# Initialize Stanza pipeline for Arabic
print("Downloading and initializing Stanza Arabic models...")
stanza.download('ar')
nlp = stanza.Pipeline('ar', processors='tokenize,pos,ner')

# 1. Data Loading and Cleaning
def load_and_preprocess(json_path, augment=False):
    """Load and preprocess the dataset with optional augmentation"""
    with open(json_path, 'r', encoding='utf-8') as f:
        data = json.load(f)
    df = pd.DataFrame(data)

    # Extract techniques and clean text
    df["techniques"] = df["labels"].apply(lambda x: [label["technique"] for label in x])
    df["text"] = df["text"].apply(lambda x: re.sub(r"@\w+|https?://\S+|#[^\s]+", "", x))

    if augment:
        # Add simple augmentation for rare classes
        rare_tech = ['Appeal to authority', 'Obfuscation', 'Repetition']
        aug_samples = []
        for _, row in df[df["techniques"].apply(lambda x: any(t in x for t in rare_tech))].iterrows():
            aug_samples.append({
                "text": re.sub(r"(\w+)\s+", r"\1 \1 ", row["text"]),  # Simple repetition
                "techniques": row["techniques"]
            })
        df = pd.concat([df, pd.DataFrame(aug_samples)])

    return df

# 2.  Arabic Text Cleaning
def clean_arabic_text(text):
    """Comprehensive cleaning and normalization of Arabic text"""
    # Normalize different forms of Arabic letters
    text = re.sub(r"[إأآا]", "ا", text)
    text = re.sub(r"[ةه]", "ه", text)
    text = re.sub(r"[ىي]", "ي", text)
    text = re.sub(r"[ؤئ]", "ء", text)
    
    # Remove diacritics and tatweel
    text = re.sub(r'[\u064B-\u065F\u0670\u0640]', '', text)
    
    # Normalize punctuation
    text = re.sub(r'[؟!]', '.', text)  # Standardize sentence endings
    text = re.sub(r'[،؛]', ',', text)  # Standardize pauses
    
    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

# 3. Linguistic Feature Extraction
def extract_linguistic_features(texts):
    """Extract POS and NER features using Stanza"""
    pos_features = []
    ner_features = []
    
    for text in texts:
        try:
            doc = nlp(text)
            
            # POS features
            pos_tags = []
            for sentence in doc.sentences:
                for word in sentence.words:
                    pos_tags.append(word.pos)
            pos_features.append(" ".join(pos_tags))
            
            # NER features
            entities = []
            for ent in doc.entities:
                entities.append(f"{ent.type}_{ent.text}")
            ner_features.append(" ".join(entities))
        except:
            
            pos_features.append("")
            ner_features.append("")
    
    return pos_features, ner_features

def create_linguistic_embeddings(pos_texts, ner_texts, n_components=16):  # Reduced from 32 to 16
    """
    Convert POS/NER texts to dense embeddings using TF-IDF + SVD
    """
    # POS embeddings
    pos_vectorizer = TfidfVectorizer(max_features=500)  # Reduced from 1000
    pos_tfidf = pos_vectorizer.fit_transform(pos_texts)
    
    # Dynamically adjust n_components for POS
    pos_n_components = min(n_components, pos_tfidf.shape[1] - 1) if pos_tfidf.shape[1] > 1 else 1
    pos_svd = TruncatedSVD(n_components=pos_n_components)
    pos_emb = pos_svd.fit_transform(pos_tfidf)
    
    # NER embeddings
    ner_vectorizer = TfidfVectorizer(max_features=250)  # Reduced from 500
    ner_tfidf = ner_vectorizer.fit_transform(ner_texts)
    
    # Dynamically adjust n_components for NER
    ner_n_components = min(n_components, ner_tfidf.shape[1] - 1) if ner_tfidf.shape[1] > 1 else 1
    ner_svd = TruncatedSVD(n_components=ner_n_components)
    ner_emb = ner_svd.fit_transform(ner_tfidf)
    
    # Pad embeddings to ensure consistent dimensionality
    if pos_emb.shape[1] < n_components:
        pos_emb = np.pad(pos_emb, ((0, 0), (0, n_components - pos_emb.shape[1])))
    if ner_emb.shape[1] < n_components:
        ner_emb = np.pad(ner_emb, ((0, 0), (0, n_components - ner_emb.shape[1])))
    
    return pos_emb, ner_emb, pos_vectorizer, ner_vectorizer, pos_svd, ner_svd


# Load datasets
print("Loading and preprocessing datasets...")
train_df = load_and_preprocess("/kaggle/input/train_ds1/all_aug_for_test.json", augment=True)
test_df = load_and_preprocess("/kaggle/input/test-ds1/task2_test_gold_label_final_mod.json")


#4.Data Preparation
all_techniques = sorted(set(tech for sublist in train_df["techniques"] for tech in sublist))
mlb = MultiLabelBinarizer(classes=all_techniques)
mlb.fit(train_df["techniques"])

train_df["cleaned_text"] = train_df["text"].apply(clean_arabic_text)
test_df["cleaned_text"] = test_df["text"].apply(clean_arabic_text)

print("Extracting linguistic features for train and test...")
train_pos_raw, train_ner_raw = extract_linguistic_features(train_df["cleaned_text"])
test_pos_raw, test_ner_raw = extract_linguistic_features(test_df["cleaned_text"])


pos_vectorizer = TfidfVectorizer(max_features=500).fit(train_pos_raw)
train_pos_tfidf = pos_vectorizer.transform(train_pos_raw)
pos_svd = TruncatedSVD(n_components=min(16, train_pos_tfidf.shape[1] - 1), random_state=42).fit(train_pos_tfidf)

ner_vectorizer = TfidfVectorizer(max_features=250).fit(train_ner_raw)
train_ner_tfidf = ner_vectorizer.transform(train_ner_raw)
ner_svd = TruncatedSVD(n_components=min(16, train_ner_tfidf.shape[1] - 1), random_state=42).fit(train_ner_tfidf)


X_train_pos = pos_svd.transform(pos_vectorizer.transform(train_pos_raw))
X_train_ner = ner_svd.transform(ner_vectorizer.transform(train_ner_raw))

X_test_pos = pos_svd.transform(pos_vectorizer.transform(test_pos_raw))
X_test_ner = ner_svd.transform(ner_vectorizer.transform(test_ner_raw))

y_train_full = mlb.transform(train_df["techniques"])
y_test = mlb.transform(test_df["techniques"])



# 5. Train/Validation Split
test_size = 0.2
split_idx = int(len(train_df) * (1 - test_size))

X_train = {
    "text": train_df["cleaned_text"][:split_idx],
    "pos_features": X_train_pos[:split_idx],
    "ner_features": X_train_ner[:split_idx]
}
y_train = y_train_full[:split_idx]

X_val = {
    "text": train_df["cleaned_text"][split_idx:],
    "pos_features": X_train_pos[split_idx:],
    "ner_features": X_train_ner[split_idx:]
}
y_val = y_train_full[split_idx:]

X_test = {
    "text": test_df["cleaned_text"],
    "pos_features": X_test_pos,
    "ner_features": X_test_ner
}
y_test = y_test


# 6. Model Components
# Initialize tokenizer and base model
print("Initializing tokenizer and base model...")
tokenizer = AutoTokenizer.from_pretrained("CAMeL-Lab/bert-base-arabic-camelbert-da-sentiment")
base_model = TFAutoModel.from_pretrained("CAMeL-Lab/bert-base-arabic-camelbert-da-sentiment")

def tokenize_dataset_with_features(texts, pos_features, ner_features):
    """Tokenize text and combine with linguistic features"""
    tokens = tokenizer(
        texts.tolist(),
        padding=True,
        truncation=True,
        max_length=128,
        return_tensors="tf"
    )
    return {
        "input_ids": tokens["input_ids"],
        "attention_mask": tokens["attention_mask"],
        "pos_features": tf.constant(pos_features),
        "ner_features": tf.constant(ner_features)
    }

# Prepare tokenized datasets
print("Tokenizing datasets...")
train_tokens = tokenize_dataset_with_features(X_train["text"], X_train["pos_features"], X_train["ner_features"])
val_tokens = tokenize_dataset_with_features(X_val["text"], X_val["pos_features"], X_val["ner_features"])
test_tokens = tokenize_dataset_with_features(X_test["text"], X_test["pos_features"], X_test["ner_features"])

# 7. Custom Loss and Metrics
class AdaptiveFocalLoss(tf.keras.losses.Loss):
    """Adaptive Focal Loss for imbalanced multi-label classification"""
    def __init__(self, class_frequencies, gamma_base=2.0, alpha=0.25, name='adaptive_focal_loss'):
        super().__init__(name=name)
        self.class_frequencies = class_frequencies
        self.gamma_base = gamma_base
        self.alpha = alpha

        # Convert to tensor with proper dtype
        self.gamma = {i: tf.constant(gamma_base * (1 + (1 - freq)), dtype=tf.float32)
                     for i, freq in enumerate(class_frequencies)}
        self.alpha_tensor = tf.constant(alpha, dtype=tf.float32)

    def call(self, y_true, y_pred):
        y_true = tf.cast(y_true, tf.float32)
        y_pred = tf.cast(y_pred, tf.float32)

        pt = tf.where(tf.equal(y_true, 1), y_pred, 1 - y_pred)
        gamma_tensor = tf.zeros_like(y_true, dtype=tf.float32)

        for i in range(y_true.shape[1]):
            gamma_tensor = tf.where(
                tf.equal(y_true[:, i:i+1], 1),
                self.gamma[i],
                gamma_tensor
            )

        loss = -self.alpha_tensor * tf.pow(1.0 - pt, gamma_tensor) * tf.math.log(pt + 1e-7)
        return tf.reduce_mean(loss)

    def get_config(self):
        config = super().get_config()
        config.update({
            'class_frequencies': self.class_frequencies.tolist(),
            'gamma_base': self.gamma_base,
            'alpha': self.alpha
        })
        return config

    @classmethod
    def from_config(cls, config):
        config['class_frequencies'] = np.array(config['class_frequencies'])
        return cls(**config)

class F1MicroScore(tf.keras.metrics.Metric):
    """Custom F1 Micro Score metric"""
    def __init__(self, name='f1_micro', **kwargs):
        super().__init__(name=name, **kwargs)
        self.precision = tf.keras.metrics.Precision(thresholds=0.5)
        self.recall = tf.keras.metrics.Recall(thresholds=0.5)

    def update_state(self, y_true, y_pred, sample_weight=None):
        self.precision.update_state(y_true, y_pred)
        self.recall.update_state(y_true, y_pred)

    def result(self):
        p = self.precision.result()
        r = self.recall.result()
        return 2 * ((p * r) / (p + r + 1e-6))

    def reset_state(self):
        self.precision.reset_state()
        self.recall.reset_state()

import tensorflow as tf
tf.keras.backend.clear_session()

# 8. Model Architecture
def build_enhanced_model_with_features():
    """Build model with linguistic feature integration"""
    # Text inputs (BERT)
    input_ids = tf.keras.layers.Input(shape=(None,), dtype=tf.int32, name="input_ids")
    attention_mask = tf.keras.layers.Input(shape=(None,), dtype=tf.int32, name="attention_mask")
    
    # Linguistic feature inputs
    pos_input = tf.keras.layers.Input(shape=(16,), name="pos_features")
    ner_input = tf.keras.layers.Input(shape=(16,), name="ner_features")

    # BERT processing
    bert_output = base_model(input_ids, attention_mask=attention_mask)
    attention = tf.keras.layers.MultiHeadAttention(num_heads=4, key_dim=64)(
        bert_output.last_hidden_state, bert_output.last_hidden_state)
    attention = tf.keras.layers.GlobalAveragePooling1D()(attention)
    
    # Process linguistic features
    pos_dense = tf.keras.layers.Dense(16, activation='swish')(pos_input)
    ner_dense = tf.keras.layers.Dense(16, activation='swish')(ner_input)
    
    # Concatenate all features
    combined = tf.keras.layers.concatenate([attention, pos_dense, ner_dense])
    
    # Feature pyramid with modified normalization
    dense1 = tf.keras.layers.Dense(256, activation='swish')(combined)
    dense1 = tf.keras.layers.BatchNormalization()(dense1)  # Changed from LayerNorm
    dense1 = tf.keras.layers.Dropout(0.3)(dense1)

    dense2 = tf.keras.layers.Dense(128, activation='swish')(dense1)
    dense2 = tf.keras.layers.BatchNormalization()(dense2)  # Changed from LayerNorm

    # Output with temperature scaling
    output = tf.keras.layers.Dense(len(all_techniques), activation="sigmoid")(dense2)
    
    return tf.keras.Model(
        inputs=[input_ids, attention_mask, pos_input, ner_input], 
        outputs=output
    )

# 9. Training Configuration
# Calculate class weights
class_counts = y_train.sum(axis=0)
total_samples = len(y_train)
class_weights = total_samples / (len(class_counts) * (1 / (class_counts + 1e-6))) 
class_weights = {i: float(w) for i, w in enumerate(class_weights)}

# Apply special weights to important classes
special_weights = {
    'Obfuscation, Intentional vagueness, Confusion': 3.0,
    'Repetition': 2.5,
    'Appeal to authority': 2.0,
    'Glittering generalities (Virtue)': 2.0
}

for tech, weight in special_weights.items():
    if tech in all_techniques:
        class_weights[all_techniques.index(tech)] = weight

# Initialize focal loss
class_freq = y_train.mean(axis=0).values.astype(np.float32)
focal_loss = AdaptiveFocalLoss(class_freq)

# Callbacks
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_auc",
    patience=15,
    mode="max",
    restore_best_weights=True
)

reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.2,
    patience=5,
    min_lr=1e-6
)

# 10. Model Training
print("Building and training model...")
model = build_enhanced_model_with_features()

model.compile(
    optimizer=tf.keras.optimizers.Adam(2e-5),
    loss=focal_loss,
    metrics=[
        tf.keras.metrics.AUC(name='auc'),
        tf.keras.metrics.Precision(name='precision'),
        tf.keras.metrics.Recall(name='recall'),
        F1MicroScore()
    ]
)

history = model.fit(
    train_tokens,
    y_train,
    validation_data=(val_tokens, y_val),
    batch_size=16,
    epochs=30,
    callbacks=[early_stopping, reduce_lr],
    class_weight=class_weights
)

# 11. Threshold Optimization
def optimize_thresholds(y_true, y_pred_probs):
    """Optimize prediction thresholds per class"""
    thresholds = []
    for i in range(y_pred_probs.shape[1]):
        if y_true[:, i].sum() == 0:
            thresholds.append(0.3)
            continue

        # Coarse search
        best_t, best_f1 = 0.3, 0
        for t in np.linspace(0.1, 0.5, 20):
            f1 = f1_score(y_true[:, i], y_pred_probs[:, i] > t, zero_division=0)
            if f1 > best_f1:
                best_t, best_f1 = t, f1

        # Fine search
        for t in np.linspace(max(0.1, best_t-0.05), min(0.5, best_t+0.05), 50):
            f1 = f1_score(y_true[:, i], y_pred_probs[:, i] > t, zero_division=0)
            if f1 > best_f1:
                best_t, best_f1 = t, f1
        thresholds.append(best_t)

    # Manual adjustments
    adjustments = {
        'Appeal to authority': 0.50,
        'Smears': 0.45,
        'Exaggeration/Minimisation': 0.45,
        'Flag-waving': 0.40,
        'Obfuscation, Intentional vagueness, Confusion': 0.35
    }
    for tech, thresh in adjustments.items():
        thresholds[all_techniques.index(tech)] = thresh

    return thresholds

val_pred_probs = model.predict(val_tokens)
optimal_thresholds = optimize_thresholds(y_val.values, val_pred_probs)

# 12. Hybrid Prediction with Rules
def apply_hybrid_rules(texts, pred_probs, thresholds):
    """Apply rule-based adjustments to model predictions"""
    arabic_rules = {
        'Appeal to authority': (
            r'(?:خبير|مختص|سلطة|توصية|بروفيسور|دكتور|عالم)(?:\s+\S+){0,3}\s+(?:قال|أكد|ذكر|صرح|أوضح|أشار)',
            0.65,
            0.6
        ),
        'Obfuscation, Intentional vagueness, Confusion': (
            r'(?:بعض|عديد|معظم|قطاع|أشخاص)\s+(?:من|منهم|لها)\b|\b(?:ربما|قد|يحتمل|يمكن|ربّما)\b|(?:قد يكون|من الممكن)',
            0.55,
            0.4
        ),
        'Repetition': (
            r'(\b\w+\b)(?:\W+\1){2,}|(?:نكرر|نؤكد|مرة أخرى|مرة أخرى)',
            0.5,
            0.4
        ),
        'Smears': (
            r'\b(?:فاسد|لص|سارق|نهب|خائن|عميل|خيانة)\b(?:\s+\S+){0,2}\s+\b(?:اتهام|اتهم|وصف|توصيف)',
            0.65,
            0.5
        ),
        'Flag-waving': (
            r'(?:وطن|أمة|شرف|انتماء|وطنية|قومية)\s+(?:فخر|تضحيات|دماء|واجب|مسؤولية)',
            0.7,
            0.5
        )
    }

    final_preds = np.zeros_like(pred_probs)
    tech_index = {tech:i for i, tech in enumerate(all_techniques)}

    for tech, (pattern, boost, max_thresh) in arabic_rules.items():
        i = tech_index[tech]
        compiled = re.compile(pattern)
        tech_thresh = min(thresholds[i] * boost, max_thresh)
        for j, text in enumerate(texts):
            if compiled.search(text):
                final_preds[j,i] = int(pred_probs[j,i] > tech_thresh)
            else:
                final_preds[j,i] = int(pred_probs[j,i] > thresholds[i])

    # Default thresholds for other classes
    for i, tech in enumerate(all_techniques):
        if tech not in arabic_rules:
            final_preds[:,i] = (pred_probs[:,i] > thresholds[i]).astype(int)

    # Ensure minimum one prediction per text
    for j in range(len(texts)):
        if not final_preds[j].any():
            final_preds[j,np.argmax(pred_probs[j])] = 1

    return final_preds

# 13. Final Evaluation
test_pred_probs = model.predict(test_tokens)
test_preds = apply_hybrid_rules(X_test["text"].tolist(), test_pred_probs, optimal_thresholds)

test_classes_present = [tech for tech in all_techniques if y_test[tech].sum() > 0]
test_class_indices = [all_techniques.index(tech) for tech in test_classes_present]

print("\nOptimized Thresholds:")
for tech, threshold in zip(all_techniques, optimal_thresholds):
    print(f"{tech:>50}: {threshold:.3f}")

print("\nEnhanced Test Performance with Linguistic Features:")
print(classification_report(
    y_test[test_classes_present],
    test_preds[:, test_class_indices],
    target_names=test_classes_present,
    zero_division=0
))

# Calculate and print F1-micro score
f1_micro = f1_score(y_test, test_preds, average='micro', zero_division=0)
print("F1-micro:", f1_micro)

# 14. Visualization
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.plot(history.history['loss'], label='Train')
plt.plot(history.history['val_loss'], label='Validation')
plt.title('Loss')
plt.legend()

plt.subplot(1, 3, 2)
plt.plot(history.history['precision'], label='Train Precision')
plt.plot(history.history['val_precision'], label='Val Precision')
plt.title('Precision')
plt.legend()

plt.subplot(1, 3, 3)
plt.plot(history.history['recall'], label='Train Recall')
plt.plot(history.history['val_recall'], label='Val Recall')
plt.title('Recall')
plt.legend()
plt.tight_layout()
plt.show()

# 15. Save Model
print("Saving model and metadata...")
model_save_path = "/models/propaganda_detector_enhanced_POS_NER"
os.makedirs(model_save_path, exist_ok=True)

# Save model
tf.keras.models.save_model(
    model,
    model_save_path,
    save_format='tf',
    include_optimizer=True
)

# Save tokenizer and metadata
tokenizer.save_pretrained(model_save_path)

metadata = {
    "mlb": mlb,
    "optimal_thresholds": optimal_thresholds,
    "all_techniques": all_techniques,
    "class_frequencies": class_freq
}
joblib.dump(metadata, f"{model_save_path}/metadata.pkl")

print("Training and evaluation complete!")
```

```python
#multilable text classification bert_pos_ner using augmented dataset
# Import all required libraries
import json
import pandas as pd
import re
import numpy as np
from sklearn.preprocessing import MultiLabelBinarizer
from transformers import AutoTokenizer, TFAutoModel
import tensorflow as tf
from sklearn.metrics import classification_report, f1_score
import matplotlib.pyplot as plt
import stanza
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from scipy.sparse import hstack
import os
import joblib

# Initialize Stanza pipeline for Arabic
print("Downloading and initializing Stanza Arabic models...")
stanza.download('ar')
nlp = stanza.Pipeline('ar', processors='tokenize,pos,ner')

# 1. Data Loading and Cleaning
def load_and_preprocess(json_path, augment=False):
    """Load and preprocess the dataset with optional augmentation"""
    with open(json_path, 'r', encoding='utf-8') as f:
        data = json.load(f)
    df = pd.DataFrame(data)

    # Extract techniques and clean text
    df["techniques"] = df["labels"].apply(lambda x: [label["technique"] for label in x])
    df["text"] = df["text"].apply(lambda x: re.sub(r"@\w+|https?://\S+|#[^\s]+", "", x))

    if augment:
        # Add simple augmentation for rare classes
        rare_tech = ['Appeal to authority', 'Obfuscation', 'Repetition']
        aug_samples = []
        for _, row in df[df["techniques"].apply(lambda x: any(t in x for t in rare_tech))].iterrows():
            aug_samples.append({
                "text": re.sub(r"(\w+)\s+", r"\1 \1 ", row["text"]),  # Simple repetition
                "techniques": row["techniques"]
            })
        df = pd.concat([df, pd.DataFrame(aug_samples)])

    return df

# 2.  Arabic Text Cleaning
def clean_arabic_text(text):
    """Comprehensive cleaning and normalization of Arabic text"""
    # Normalize different forms of Arabic letters
    text = re.sub(r"[إأآا]", "ا", text)
    text = re.sub(r"[ةه]", "ه", text)
    text = re.sub(r"[ىي]", "ي", text)
    text = re.sub(r"[ؤئ]", "ء", text)
    
    # Remove diacritics and tatweel
    text = re.sub(r'[\u064B-\u065F\u0670\u0640]', '', text)
    
    # Normalize punctuation
    text = re.sub(r'[؟!]', '.', text)  # Standardize sentence endings
    text = re.sub(r'[،؛]', ',', text)  # Standardize pauses
    
    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

# 3. Linguistic Feature Extraction
def extract_linguistic_features(texts):
    """Extract POS and NER features using Stanza"""
    pos_features = []
    ner_features = []
    
    for text in texts:
        try:
            doc = nlp(text)
            
            # POS features
            pos_tags = []
            for sentence in doc.sentences:
                for word in sentence.words:
                    pos_tags.append(word.pos)
            pos_features.append(" ".join(pos_tags))
            
            # NER features
            entities = []
            for ent in doc.entities:
                entities.append(f"{ent.type}_{ent.text}")
            ner_features.append(" ".join(entities))
        except:
            
            pos_features.append("")
            ner_features.append("")
    
    return pos_features, ner_features

def create_linguistic_embeddings(pos_texts, ner_texts, n_components=16):  # Reduced from 32 to 16
    """
    Convert POS/NER texts to dense embeddings using TF-IDF + SVD
    """
    # POS embeddings
    pos_vectorizer = TfidfVectorizer(max_features=500)  # Reduced from 1000
    pos_tfidf = pos_vectorizer.fit_transform(pos_texts)
    
    # Dynamically adjust n_components for POS
    pos_n_components = min(n_components, pos_tfidf.shape[1] - 1) if pos_tfidf.shape[1] > 1 else 1
    pos_svd = TruncatedSVD(n_components=pos_n_components)
    pos_emb = pos_svd.fit_transform(pos_tfidf)
    
    # NER embeddings
    ner_vectorizer = TfidfVectorizer(max_features=250)  # Reduced from 500
    ner_tfidf = ner_vectorizer.fit_transform(ner_texts)
    
    # Dynamically adjust n_components for NER
    ner_n_components = min(n_components, ner_tfidf.shape[1] - 1) if ner_tfidf.shape[1] > 1 else 1
    ner_svd = TruncatedSVD(n_components=ner_n_components)
    ner_emb = ner_svd.fit_transform(ner_tfidf)
    
    # Pad embeddings to ensure consistent dimensionality
    if pos_emb.shape[1] < n_components:
        pos_emb = np.pad(pos_emb, ((0, 0), (0, n_components - pos_emb.shape[1])))
    if ner_emb.shape[1] < n_components:
        ner_emb = np.pad(ner_emb, ((0, 0), (0, n_components - ner_emb.shape[1])))
    
    return pos_emb, ner_emb, pos_vectorizer, ner_vectorizer, pos_svd, ner_svd


# Load datasets
print("Loading and preprocessing datasets...")
train_df = load_and_preprocess("/kaggle/input/original_train_ds2/task2_train.json", augment=True)
test_df = load_and_preprocess("/kaggle/input/test-ds1/task2_test_gold_label_final_mod.json")


#4.Data Preparation
all_techniques = sorted(set(tech for sublist in train_df["techniques"] for tech in sublist))
mlb = MultiLabelBinarizer(classes=all_techniques)
mlb.fit(train_df["techniques"])

train_df["cleaned_text"] = train_df["text"].apply(clean_arabic_text)
test_df["cleaned_text"] = test_df["text"].apply(clean_arabic_text)

print("Extracting linguistic features for train and test...")
train_pos_raw, train_ner_raw = extract_linguistic_features(train_df["cleaned_text"])
test_pos_raw, test_ner_raw = extract_linguistic_features(test_df["cleaned_text"])


pos_vectorizer = TfidfVectorizer(max_features=500).fit(train_pos_raw)
train_pos_tfidf = pos_vectorizer.transform(train_pos_raw)
pos_svd = TruncatedSVD(n_components=min(16, train_pos_tfidf.shape[1] - 1), random_state=42).fit(train_pos_tfidf)

ner_vectorizer = TfidfVectorizer(max_features=250).fit(train_ner_raw)
train_ner_tfidf = ner_vectorizer.transform(train_ner_raw)
ner_svd = TruncatedSVD(n_components=min(16, train_ner_tfidf.shape[1] - 1), random_state=42).fit(train_ner_tfidf)


X_train_pos = pos_svd.transform(pos_vectorizer.transform(train_pos_raw))
X_train_ner = ner_svd.transform(ner_vectorizer.transform(train_ner_raw))

X_test_pos = pos_svd.transform(pos_vectorizer.transform(test_pos_raw))
X_test_ner = ner_svd.transform(ner_vectorizer.transform(test_ner_raw))

y_train_full = mlb.transform(train_df["techniques"])
y_test = mlb.transform(test_df["techniques"])



# 5. Train/Validation Split
test_size = 0.2
split_idx = int(len(train_df) * (1 - test_size))

X_train = {
    "text": train_df["cleaned_text"][:split_idx],
    "pos_features": X_train_pos[:split_idx],
    "ner_features": X_train_ner[:split_idx]
}
y_train = y_train_full[:split_idx]

X_val = {
    "text": train_df["cleaned_text"][split_idx:],
    "pos_features": X_train_pos[split_idx:],
    "ner_features": X_train_ner[split_idx:]
}
y_val = y_train_full[split_idx:]

X_test = {
    "text": test_df["cleaned_text"],
    "pos_features": X_test_pos,
    "ner_features": X_test_ner
}
y_test = y_test


# 6. Model Components
# Initialize tokenizer and base model
print("Initializing tokenizer and base model...")
tokenizer = AutoTokenizer.from_pretrained("CAMeL-Lab/bert-base-arabic-camelbert-da-sentiment")
base_model = TFAutoModel.from_pretrained("CAMeL-Lab/bert-base-arabic-camelbert-da-sentiment")

def tokenize_dataset_with_features(texts, pos_features, ner_features):
    """Tokenize text and combine with linguistic features"""
    tokens = tokenizer(
        texts.tolist(),
        padding=True,
        truncation=True,
        max_length=128,
        return_tensors="tf"
    )
    return {
        "input_ids": tokens["input_ids"],
        "attention_mask": tokens["attention_mask"],
        "pos_features": tf.constant(pos_features),
        "ner_features": tf.constant(ner_features)
    }

# Prepare tokenized datasets
print("Tokenizing datasets...")
train_tokens = tokenize_dataset_with_features(X_train["text"], X_train["pos_features"], X_train["ner_features"])
val_tokens = tokenize_dataset_with_features(X_val["text"], X_val["pos_features"], X_val["ner_features"])
test_tokens = tokenize_dataset_with_features(X_test["text"], X_test["pos_features"], X_test["ner_features"])

# 7. Custom Loss and Metrics
class AdaptiveFocalLoss(tf.keras.losses.Loss):
    """Adaptive Focal Loss for imbalanced multi-label classification"""
    def __init__(self, class_frequencies, gamma_base=2.0, alpha=0.25, name='adaptive_focal_loss'):
        super().__init__(name=name)
        self.class_frequencies = class_frequencies
        self.gamma_base = gamma_base
        self.alpha = alpha

        # Convert to tensor with proper dtype
        self.gamma = {i: tf.constant(gamma_base * (1 + (1 - freq)), dtype=tf.float32)
                     for i, freq in enumerate(class_frequencies)}
        self.alpha_tensor = tf.constant(alpha, dtype=tf.float32)

    def call(self, y_true, y_pred):
        y_true = tf.cast(y_true, tf.float32)
        y_pred = tf.cast(y_pred, tf.float32)

        pt = tf.where(tf.equal(y_true, 1), y_pred, 1 - y_pred)
        gamma_tensor = tf.zeros_like(y_true, dtype=tf.float32)

        for i in range(y_true.shape[1]):
            gamma_tensor = tf.where(
                tf.equal(y_true[:, i:i+1], 1),
                self.gamma[i],
                gamma_tensor
            )

        loss = -self.alpha_tensor * tf.pow(1.0 - pt, gamma_tensor) * tf.math.log(pt + 1e-7)
        return tf.reduce_mean(loss)

    def get_config(self):
        config = super().get_config()
        config.update({
            'class_frequencies': self.class_frequencies.tolist(),
            'gamma_base': self.gamma_base,
            'alpha': self.alpha
        })
        return config

    @classmethod
    def from_config(cls, config):
        config['class_frequencies'] = np.array(config['class_frequencies'])
        return cls(**config)

class F1MicroScore(tf.keras.metrics.Metric):
    """Custom F1 Micro Score metric"""
    def __init__(self, name='f1_micro', **kwargs):
        super().__init__(name=name, **kwargs)
        self.precision = tf.keras.metrics.Precision(thresholds=0.5)
        self.recall = tf.keras.metrics.Recall(thresholds=0.5)

    def update_state(self, y_true, y_pred, sample_weight=None):
        self.precision.update_state(y_true, y_pred)
        self.recall.update_state(y_true, y_pred)

    def result(self):
        p = self.precision.result()
        r = self.recall.result()
        return 2 * ((p * r) / (p + r + 1e-6))

    def reset_state(self):
        self.precision.reset_state()
        self.recall.reset_state()

import tensorflow as tf
tf.keras.backend.clear_session()

# 8. Model Architecture
def build_enhanced_model_with_features():
    """Build model with linguistic feature integration"""
    # Text inputs (BERT)
    input_ids = tf.keras.layers.Input(shape=(None,), dtype=tf.int32, name="input_ids")
    attention_mask = tf.keras.layers.Input(shape=(None,), dtype=tf.int32, name="attention_mask")
    
    # Linguistic feature inputs
    pos_input = tf.keras.layers.Input(shape=(16,), name="pos_features")
    ner_input = tf.keras.layers.Input(shape=(16,), name="ner_features")

    # BERT processing
    bert_output = base_model(input_ids, attention_mask=attention_mask)
    attention = tf.keras.layers.MultiHeadAttention(num_heads=4, key_dim=64)(
        bert_output.last_hidden_state, bert_output.last_hidden_state)
    attention = tf.keras.layers.GlobalAveragePooling1D()(attention)
    
    # Process linguistic features
    pos_dense = tf.keras.layers.Dense(16, activation='swish')(pos_input)
    ner_dense = tf.keras.layers.Dense(16, activation='swish')(ner_input)
    
    # Concatenate all features
    combined = tf.keras.layers.concatenate([attention, pos_dense, ner_dense])
    
    # Feature pyramid with modified normalization
    dense1 = tf.keras.layers.Dense(256, activation='swish')(combined)
    dense1 = tf.keras.layers.BatchNormalization()(dense1)  # Changed from LayerNorm
    dense1 = tf.keras.layers.Dropout(0.3)(dense1)

    dense2 = tf.keras.layers.Dense(128, activation='swish')(dense1)
    dense2 = tf.keras.layers.BatchNormalization()(dense2)  # Changed from LayerNorm

    # Output with temperature scaling
    output = tf.keras.layers.Dense(len(all_techniques), activation="sigmoid")(dense2)
    
    return tf.keras.Model(
        inputs=[input_ids, attention_mask, pos_input, ner_input], 
        outputs=output
    )

# 9. Training Configuration
# Calculate class weights
class_counts = y_train.sum(axis=0)
total_samples = len(y_train)
class_weights = total_samples / (len(class_counts) * (1 / (class_counts + 1e-6))) 
class_weights = {i: float(w) for i, w in enumerate(class_weights)}

# Apply special weights to important classes
special_weights = {
    'Obfuscation, Intentional vagueness, Confusion': 3.0,
    'Repetition': 2.5,
    'Appeal to authority': 2.0,
    'Glittering generalities (Virtue)': 2.0
}

for tech, weight in special_weights.items():
    if tech in all_techniques:
        class_weights[all_techniques.index(tech)] = weight

# Initialize focal loss
class_freq = y_train.mean(axis=0).values.astype(np.float32)
focal_loss = AdaptiveFocalLoss(class_freq)

# Callbacks
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_auc",
    patience=15,
    mode="max",
    restore_best_weights=True
)

reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.2,
    patience=5,
    min_lr=1e-6
)

# 10. Model Training
print("Building and training model...")
model = build_enhanced_model_with_features()

model.compile(
    optimizer=tf.keras.optimizers.Adam(2e-5),
    loss=focal_loss,
    metrics=[
        tf.keras.metrics.AUC(name='auc'),
        tf.keras.metrics.Precision(name='precision'),
        tf.keras.metrics.Recall(name='recall'),
        F1MicroScore()
    ]
)

history = model.fit(
    train_tokens,
    y_train,
    validation_data=(val_tokens, y_val),
    batch_size=16,
    epochs=30,
    callbacks=[early_stopping, reduce_lr],
    class_weight=class_weights
)

# 11. Threshold Optimization
def optimize_thresholds(y_true, y_pred_probs):
    """Optimize prediction thresholds per class"""
    thresholds = []
    for i in range(y_pred_probs.shape[1]):
        if y_true[:, i].sum() == 0:
            thresholds.append(0.3)
            continue

        # Coarse search
        best_t, best_f1 = 0.3, 0
        for t in np.linspace(0.1, 0.5, 20):
            f1 = f1_score(y_true[:, i], y_pred_probs[:, i] > t, zero_division=0)
            if f1 > best_f1:
                best_t, best_f1 = t, f1

        # Fine search
        for t in np.linspace(max(0.1, best_t-0.05), min(0.5, best_t+0.05), 50):
            f1 = f1_score(y_true[:, i], y_pred_probs[:, i] > t, zero_division=0)
            if f1 > best_f1:
                best_t, best_f1 = t, f1
        thresholds.append(best_t)

    # Manual adjustments
    adjustments = {
        'Appeal to authority': 0.50,
        'Smears': 0.45,
        'Exaggeration/Minimisation': 0.45,
        'Flag-waving': 0.40,
        'Obfuscation, Intentional vagueness, Confusion': 0.35
    }
    for tech, thresh in adjustments.items():
        thresholds[all_techniques.index(tech)] = thresh

    return thresholds

val_pred_probs = model.predict(val_tokens)
optimal_thresholds = optimize_thresholds(y_val.values, val_pred_probs)

# 12. Hybrid Prediction with Rules
def apply_hybrid_rules(texts, pred_probs, thresholds):
    """Apply rule-based adjustments to model predictions"""
    arabic_rules = {
        'Appeal to authority': (
            r'(?:خبير|مختص|سلطة|توصية|بروفيسور|دكتور|عالم)(?:\s+\S+){0,3}\s+(?:قال|أكد|ذكر|صرح|أوضح|أشار)',
            0.65,
            0.6
        ),
        'Obfuscation, Intentional vagueness, Confusion': (
            r'(?:بعض|عديد|معظم|قطاع|أشخاص)\s+(?:من|منهم|لها)\b|\b(?:ربما|قد|يحتمل|يمكن|ربّما)\b|(?:قد يكون|من الممكن)',
            0.55,
            0.4
        ),
        'Repetition': (
            r'(\b\w+\b)(?:\W+\1){2,}|(?:نكرر|نؤكد|مرة أخرى|مرة أخرى)',
            0.5,
            0.4
        ),
        'Smears': (
            r'\b(?:فاسد|لص|سارق|نهب|خائن|عميل|خيانة)\b(?:\s+\S+){0,2}\s+\b(?:اتهام|اتهم|وصف|توصيف)',
            0.65,
            0.5
        ),
        'Flag-waving': (
            r'(?:وطن|أمة|شرف|انتماء|وطنية|قومية)\s+(?:فخر|تضحيات|دماء|واجب|مسؤولية)',
            0.7,
            0.5
        )
    }

    final_preds = np.zeros_like(pred_probs)
    tech_index = {tech:i for i, tech in enumerate(all_techniques)}

    for tech, (pattern, boost, max_thresh) in arabic_rules.items():
        i = tech_index[tech]
        compiled = re.compile(pattern)
        tech_thresh = min(thresholds[i] * boost, max_thresh)
        for j, text in enumerate(texts):
            if compiled.search(text):
                final_preds[j,i] = int(pred_probs[j,i] > tech_thresh)
            else:
                final_preds[j,i] = int(pred_probs[j,i] > thresholds[i])

    # Default thresholds for other classes
    for i, tech in enumerate(all_techniques):
        if tech not in arabic_rules:
            final_preds[:,i] = (pred_probs[:,i] > thresholds[i]).astype(int)

    # Ensure minimum one prediction per text
    for j in range(len(texts)):
        if not final_preds[j].any():
            final_preds[j,np.argmax(pred_probs[j])] = 1

    return final_preds

# 13. Final Evaluation
test_pred_probs = model.predict(test_tokens)
test_preds = apply_hybrid_rules(X_test["text"].tolist(), test_pred_probs, optimal_thresholds)

test_classes_present = [tech for tech in all_techniques if y_test[tech].sum() > 0]
test_class_indices = [all_techniques.index(tech) for tech in test_classes_present]

print("\nOptimized Thresholds:")
for tech, threshold in zip(all_techniques, optimal_thresholds):
    print(f"{tech:>50}: {threshold:.3f}")

print("\nEnhanced Test Performance with Linguistic Features:")
print(classification_report(
    y_test[test_classes_present],
    test_preds[:, test_class_indices],
    target_names=test_classes_present,
    zero_division=0
))

# Calculate and print F1-micro score
f1_micro = f1_score(y_test, test_preds, average='micro', zero_division=0)
print("F1-micro:", f1_micro)

# 14. Visualization
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.plot(history.history['loss'], label='Train')
plt.plot(history.history['val_loss'], label='Validation')
plt.title('Loss')
plt.legend()

plt.subplot(1, 3, 2)
plt.plot(history.history['precision'], label='Train Precision')
plt.plot(history.history['val_precision'], label='Val Precision')
plt.title('Precision')
plt.legend()

plt.subplot(1, 3, 3)
plt.plot(history.history['recall'], label='Train Recall')
plt.plot(history.history['val_recall'], label='Val Recall')
plt.title('Recall')
plt.legend()
plt.tight_layout()
plt.show()

# 15. Save Model
print("Saving model and metadata...")
model_save_path = "/models/propaganda_detector_enhanced_POS_NER"
os.makedirs(model_save_path, exist_ok=True)

# Save model
tf.keras.models.save_model(
    model,
    model_save_path,
    save_format='tf',
    include_optimizer=True
)

# Save tokenizer and metadata
tokenizer.save_pretrained(model_save_path)

metadata = {
    "mlb": mlb,
    "optimal_thresholds": optimal_thresholds,
    "all_techniques": all_techniques,
    "class_frequencies": class_freq
}
joblib.dump(metadata, f"{model_save_path}/metadata.pkl")

print("Training and evaluation complete!")
```
